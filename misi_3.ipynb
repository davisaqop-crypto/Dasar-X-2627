```python
print("=== KASIR PINTAR KANTIN SMK RPL ===")
nama_barang = input("Nama Makanan/Minuman : ")
harga = int(input("Harga Satuan (Rp)    : "))
jumlah = int(input("Jumlah Beli          : "))

total_belanja = harga * jumlah

# 1. Cek Diskon Belanja
if total_belanja >= 50000:
    diskon = 5000
    print("🎉 Selamat! Anda mendapat diskon Rp5.000")
else:
    diskon = 0

bayar_akhir = total_belanja - diskon
print(f"Total yang harus dibayar: Rp{bayar_akhir}")
print("-----------------------------------")
```
