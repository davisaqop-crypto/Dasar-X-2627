Membuat program menghitung keliling persegi panjang

```python
%%writefile keliling.cpp
#include <iostream>
using namespace std;

int main() {
    int panjang, lebar, keliling;

    cout << "Masukkan panjang : ";
    cin >> panjang;
    cout << "Masukkan lebar   : ";
    cin >> lebar;

    keliling = 2 * (panjang + lebar);

    cout << "Keliling : " << keliling << endl;

    return 0;
}
```

```python
!g++ keliling.cpp -o main
```
