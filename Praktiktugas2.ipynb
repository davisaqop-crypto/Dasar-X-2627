```python
%%writefile kuadrat.cpp
#include <iostream>
using namespace std;

int main() {
    float angka, kuadrat;

    cout << "Masukkan sebuah angka: ";
    cin >> angka;

    kuadrat = angka * angka;

    cout << "Kuadrat dari " << angka << " adalah " << kuadrat << endl;

    return 0;
}
```
